```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
features_model_1 = [
    "network_size",
    "slice_type",
    "flows",
    "avg_delay",
    "jitter",
    "loss_rate",
    "throughput",
    "avg_bw_demand",
    "total_bw",
    "total_power",
    "total_cpu",
    "simulation_time"
]

target_model_1 = "allocated_bw"

df_model_1 = df[features_model_1 + [target_model_1]].copy()

df_model_1.head()
```

```python
X_1 = df_model_1[features_model_1]
y_1 = df_model_1[target_model_1]
```

```python
X_1_encoded = pd.get_dummies(X_1, columns=["slice_type"], drop_first=False)
```

```python
X_train_1, X_test_1, y_train_1, y_test_1 = train_test_split(
    X_1_encoded,
    y_1,
    test_size=0.2,
    random_state=42,
    stratify=X_1["slice_type"]
)

print("Размер X_train:", X_train_1.shape)
print("Размер X_test:", X_test_1.shape)
print("Размер y_train:", y_train_1.shape)
print("Размер y_test:", y_test_1.shape)
```

```python
scaler_1 = StandardScaler()

X_train_1_scaled = scaler_1.fit_transform(X_train_1)
X_test_1_scaled = scaler_1.transform(X_test_1)
```

```python
df_train_baseline_1 = X_1.loc[X_train_1.index].copy()
df_test_baseline_1 = X_1.loc[X_test_1.index].copy()

df_train_baseline_1["allocated_bw"] = y_train_1

slice_mean_allocated_bw = df_train_baseline_1.groupby("slice_type")["allocated_bw"].mean()

slice_mean_allocated_bw
```

```python
y_pred_baseline_1 = df_test_baseline_1["slice_type"].map(slice_mean_allocated_bw)

y_pred_baseline_1.head()
```

## Сравнение стратегий распределения bandwidth

В работе рассматриваются три стратегии распределения полосы пропускания между сетевыми срезами 5G NR.

**Demand-based Allocation** — распределение ресурса пропорционально среднему спросу на bandwidth (`avg_bw_demand`). Чем выше потребность среза в полосе пропускания, тем большую долю общего ресурса `total_bw` он получает.

**Priority-based Allocation** — распределение ресурса с учётом приоритета типа среза. Для URLLC задаётся наибольший приоритет, так как данный срез чувствителен к задержкам и потерям. eMBB получает средний приоритет из-за требований к высокой пропускной способности, а mMTC получает базовый приоритет.

**ML-based Allocation** — распределение bandwidth на основе обученной модели машинного обучения. Модель прогнозирует значение `allocated_bw` с учётом состояния сети, QoS-параметров, нагрузки и доступных ресурсов. Такой подход позволяет адаптировать распределение полосы пропускания к текущим условиям сети.

Сравнение стратегий выполняется по нескольким критериям: соблюдение ограничения `sum(allocated_bw) <= total_bw`, удовлетворение спроса на bandwidth, прогнозируемый `reward`, доля ресурсных нарушений и показатель изоляции срезов.

```python
embb_data = simulation_results[simulation_results["slice_type"] == "eMBB"]
urllc_data = simulation_results[simulation_results["slice_type"] == "URLLC"]
mmtc_data = simulation_results[simulation_results["slice_type"] == "mMTC"]

embb_throughput_threshold = embb_data["throughput"].quantile(0.25)

urllc_delay_threshold = urllc_data["avg_delay"].quantile(0.75)
urllc_jitter_threshold = urllc_data["jitter"].quantile(0.75)
urllc_loss_threshold = urllc_data["loss_rate"].quantile(0.75)

mmtc_loss_threshold = mmtc_data["loss_rate"].quantile(0.75)
mmtc_flows_threshold = mmtc_data["flows"].quantile(0.50)

print("eMBB throughput threshold:", embb_throughput_threshold)
print("URLLC delay threshold:", urllc_delay_threshold)
print("URLLC jitter threshold:", urllc_jitter_threshold)
print("URLLC loss threshold:", urllc_loss_threshold)
print("mMTC loss threshold:", mmtc_loss_threshold)
print("mMTC flows threshold:", mmtc_flows_threshold)
```

```python
simulation_results["qos_violation_embb"] = False

mask_embb = simulation_results["slice_type"] == "eMBB"

simulation_results.loc[mask_embb, "qos_violation_embb"] = (
    simulation_results.loc[mask_embb, "throughput"] < embb_throughput_threshold
)

simulation_results.loc[mask_embb, [
    "slice_type",
    "throughput",
    "qos_violation_embb"
]].head()
```

```python
simulation_results["qos_violation_urllc"] = False

mask_urllc = simulation_results["slice_type"] == "URLLC"

simulation_results.loc[mask_urllc, "qos_violation_urllc"] = (
    (simulation_results.loc[mask_urllc, "avg_delay"] > urllc_delay_threshold)
    | (simulation_results.loc[mask_urllc, "jitter"] > urllc_jitter_threshold)
    | (simulation_results.loc[mask_urllc, "loss_rate"] > urllc_loss_threshold)
)

simulation_results.loc[mask_urllc, [
    "slice_type",
    "avg_delay",
    "jitter",
    "loss_rate",
    "qos_violation_urllc"
]].head()
```

```python
simulation_results["qos_violation_mmtc"] = False

mask_mmtc = simulation_results["slice_type"] == "mMTC"

simulation_results.loc[mask_mmtc, "qos_violation_mmtc"] = (
    (simulation_results.loc[mask_mmtc, "flows"] >= mmtc_flows_threshold)
    & (simulation_results.loc[mask_mmtc, "loss_rate"] > mmtc_loss_threshold)
)

simulation_results.loc[mask_mmtc, [
    "slice_type",
    "flows",
    "loss_rate",
    "qos_violation_mmtc"
]].head()
```

```python
simulation_results["qos_violation"] = (
    simulation_results["qos_violation_embb"]
    | simulation_results["qos_violation_urllc"]
    | simulation_results["qos_violation_mmtc"]
)

simulation_results[[
    "slice_type",
    "avg_delay",
    "jitter",
    "loss_rate",
    "throughput",
    "flows",
    "qos_violation"
]].head()
```

```python
def predict_reward_for_strategy(data, bandwidth_column):
    temp = data.copy()
    temp["allocated_bw"] = temp[bandwidth_column]
    temp["utilization_bw"] = temp["allocated_bw"] / temp["total_bw"]
    temp["utilization_bw"] = temp["utilization_bw"].replace([np.inf, -np.inf], np.nan)
    temp["utilization_bw"] = temp["utilization_bw"].fillna(0)

    X_temp = temp[features_model_2].copy()
    X_temp_encoded = pd.get_dummies(X_temp, columns=["slice_type"], drop_first=False)
    X_temp_encoded = X_temp_encoded.reindex(columns=X_train_2.columns, fill_value=0)

    X_temp_scaled = scaler_2.transform(X_temp_encoded)

    if best_model_name_2 == "Baseline by slice_type":
        y_pred_temp = X_temp["slice_type"].map(slice_mean_reward).values
    else:
        y_pred_temp = best_model_2.predict(X_temp_scaled)

    return y_pred_temp
```

```python
strategy_comparison = bandwidth_strategy_summary.merge(
    satisfaction_summary,
    on="strategy",
    how="left"
)

strategy_comparison = strategy_comparison.merge(
    reward_strategy_summary,
    on="strategy",
    how="left"
)

strategy_comparison = strategy_comparison.merge(
    resource_violation_summary,
    on="strategy",
    how="left"
)

strategy_comparison["Isolation Score"] = 1 - strategy_comparison["Resource Violation Rate"]

strategy_comparison
```

```python
strategy_comparison_sorted = strategy_comparison.sort_values(
    "Average Predicted Reward",
    ascending=False
)

strategy_comparison_sorted
```

```python
strategy_score = strategy_comparison.copy()

strategy_score["reward_norm"] = (
    strategy_score["Average Predicted Reward"] - strategy_score["Average Predicted Reward"].min()
) / (
    strategy_score["Average Predicted Reward"].max() - strategy_score["Average Predicted Reward"].min()
)

strategy_score["satisfaction_norm"] = (
    strategy_score["Average Demand Satisfaction"] - strategy_score["Average Demand Satisfaction"].min()
) / (
    strategy_score["Average Demand Satisfaction"].max() - strategy_score["Average Demand Satisfaction"].min()
)

strategy_score["isolation_norm"] = (
    strategy_score["Isolation Score"] - strategy_score["Isolation Score"].min()
) / (
    strategy_score["Isolation Score"].max() - strategy_score["Isolation Score"].min()
)

strategy_score["bandwidth_violation_norm"] = 1 - (
    strategy_score["Bandwidth Violation Rate"] - strategy_score["Bandwidth Violation Rate"].min()
) / (
    strategy_score["Bandwidth Violation Rate"].max() - strategy_score["Bandwidth Violation Rate"].min()
)

strategy_score = strategy_score.fillna(0)

strategy_score["Final Strategy Score"] = (
    0.4 * strategy_score["reward_norm"]
    + 0.25 * strategy_score["satisfaction_norm"]
    + 0.25 * strategy_score["isolation_norm"]
    + 0.1 * strategy_score["bandwidth_violation_norm"]
)

strategy_score = strategy_score.sort_values("Final Strategy Score", ascending=False)

strategy_score
```
